# Helper Functions

```python
def ReLU(Z):
    return np.maximum(Z, 0)

def softmax(Z, axis):
    # Z = Z - np.max(matrix, axis=1, keepdims=True)
    # return np.exp(Z) / np.sum(np.exp(Z), axis=1, keepdims=True)
    max_val = np.max(Z, axis=axis, keepdims=True)
    exponents = np.exp(Z - max_val)
    return exponents / np.sum(exponents, axis=axis, keepdims=True)

def one_hot(Y):
    Y = Y.astype(int)
    # one_hot_Y = np.zeros(( Y.size, (np.max(Y) - np.min(Y) + 1) ))   # = array with dimensions: [ (n) , (# of classes, which is 10) ]
    one_hot_Y = np.zeros(( Y.size, len(np.unique(Y)) ))   # = array with dimensions: [ (n) , (# of classes, which is 10) ]
    one_hot_Y[np.arange(Y.size), Y - np.min(Y)] = 1
    return one_hot_Y
```

```python
class NN():

    def __init__(self, X_train: np.ndarray, Y_train: np.ndarray, X_test: np.ndarray, Y_test: np.ndarray, hidden_layer_sizes: np.ndarray):
        # data should be: rows=features, cols=datapoints
        self.X_train = X_train
        # print (X_train.shape)
        self.Y_train = Y_train
        self.X_test = X_test
        self.Y_test = Y_test
        self.num_train_data_points = X_train.shape[1]
        self.num_test_data_points = X_test.shape[1]
        self.num_data_points = self.num_train_data_points + self.num_test_data_points
        self.num_hidden_layers = len(hidden_layer_sizes) # one number to indicate # of hidden layers (excluding input & output layers)
        self.total_num_layers = self.num_hidden_layers + 2
        self.layer_sizes = np.append(np.append(X_train.shape[0], hidden_layer_sizes), len(np.unique(Y_train))) # array containing every layer size (including input & output layer)
        # self.rng = np.default_rng(seed=42)
        # np.random.seed(42)

        self.init_network()

    def init_network(self):
        self.weight_matrices = [] # len = L - 1 (L = # all layers)
        self.bias_vectors = [] # len = L - 1 (L = # all layers)
        # print ("init")
        for index in range(1, self.total_num_layers):
            # print ("index", index)
            # print (self.weight_matrices.shape)
            n_out = int(self.layer_sizes[index])
            n_in = int(self.layer_sizes[index-1])
            # W1 = self.rng.uniform(low=-1.0, high=1.0, size=(self.neurons_per_layer, np.shape(X_train)[0]))
            # W = np.random.randn(n_out, n_in) # shape = [output x input]
            np.random.seed(42)
            W = np.random.randn(n_out, n_in) * np.sqrt(2.0 / n_in) # shape = [output x input]
            b = np.zeros((n_out, 1))
            # b = np.full(n_out, 1.0).reshape(n_out, 1)
            
            self.weight_matrices.append(W)
            self.bias_vectors.append(b)

            
            '''n_in = layer_sizes[i]
            n_out = layer_sizes[i+1]
            
            # He Initialization (recommended for ReLU activation)
            w = np.random.randn(n_out, n_in) * np.sqrt(2.0 / n_in)
            # Biases initialized to zero 2D column vectors
            b = np.zeros((n_out, 1))'''

    def train(self, epochs, learning_rate):
        self.epochs = epochs
        self.learning_rate = learning_rate
        self.accuracies_over_all_epochs = np.array([])
        for epoch in range(epochs):

            # print ("---------------------------------")
            # print (f"epoch: {epoch}\n")
            
            self.forward_prop()
            self.back_prop()

            # if epoch % 10 == 0:
            #     print ("---------------------------------")
            #     print (f"epoch: {epoch}")
            #     self.predict()
            #     print ()
            if epoch % 10 == 0:
                print ("---------------------------------")
                print (f"epoch: {epoch}")
                self.predict(should_print=True)
                print ()
            else:
                self.predict(should_print=False)

        print (f"epoch # and accuracy of highest accuracy epoch: {np.argmax(self.accuracies_over_all_epochs)}, {self.accuracies_over_all_epochs[np.argmax(self.accuracies_over_all_epochs)]}")
        



    # Core Functions
    
    def forward_prop(self):

        self.weighted_sum_matrices = [self.X_train] # all of the Z matrices (used for backprop)
        self.activations = [self.X_train] # training_set + activations
        
        # for (index, weight_matrix) in enumerate(self.weight_matrices):
        for index in range(self.num_hidden_layers):
            Z = self.weight_matrices[index] @ self.activations[index]
            Z += self.bias_vectors[index] # Z = linear sum --> (W @ X) + b
            # print (f"Z, index {index}\n{Z}")
            self.weighted_sum_matrices.append(Z)
            A = NN.ReLU(Z) # A = activation of Z
            # print (f"A, index {index}\n{A}")
            self.activations.append(A)

        # for the output layer
        # Z = self.weight_matrices[-1] @ self.forward_prop_inputs[-1]
        Z = self.weight_matrices[-1] @ self.activations[-1]
        Z += self.bias_vectors[-1]
        # self.weighted_sum_matrices.append(Z) # not needed?

        # print (f"Output Z\n{Z}")
        
        self.output_probability_matrix = NN.softmax(Z, axis=0).T # softmax activation instead of ReLU, shape: rows=datapoints , cols=features
        # print (f"output_probability_matrix\n{self.output_probability_matrix}")
        # self.activations.append(output_probability_matrix)

        # print ("finished forward prop")
        # print ("len(self.weighted_sum_matrices)", len(self.weighted_sum_matrices))
        # print ("len(self.activations)", len(self.activations), "\n")


        # each matrix in self.forward_prop_inputs has shape: [output_neurons x datapoints]

    def back_prop(self):
        one_hot_Y = NN.one_hot(self.Y_train) # shape: [n x c] --> rows=each datapoint , cols=each class possibility (ie. 0, 1, 2)

        assert self.output_probability_matrix.shape == one_hot_Y.shape
        error_gradient = self.output_probability_matrix - one_hot_Y # shape = [n x c]
        error_gradient = error_gradient.T # shape = [c x n]

        # print ("PRE back prop weight matrices")
        # for i in self.weight_matrices:
        #     print (i, "\n")
        # print ("\n")

        for index in range(self.total_num_layers - 2, -1, -1):
            
            # print ("backprop index", index, "\n")
            # # print ("error_gradient shape", error_gradient.shape)
            # print ("error gradient shape\n", error_gradient.shape, "\n")
            # print ("error gradient\n", error_gradient, "\n")
            
            # print ("err sum", np.sum(error_gradient, axis=1))
            # print ("err sum reshaped", np.sum(error_gradient, axis=1).T.shape)

            weighted_sum_matrix = self.weighted_sum_matrices[index]
            # activation_matrix = self.forward_prop_inputs[index].T
            activation_matrix = self.activations[index].T

            # print ("weighted_sum_matrix\n", weighted_sum_matrix, "\n")
            # print ("activation_matrix\n", activation_matrix, "\n")

            # print ("weight matrix value for sub\n", ((self.learning_rate) * (np.dot(error_gradient, activation_matrix))) / self.num_train_data_points, "\n")
            self.weight_matrices[index] -= ((self.learning_rate) * (np.dot(error_gradient, activation_matrix))) / self.num_train_data_points
            # print ("self.bias_vectors[index] shape", self.bias_vectors[index].shape)
            # print ("bias error value for subtraction shape", ((self.learning_rate) * (np.sum(error_gradient, axis=1) / self.num_train_data_points)).shape)
            self.bias_vectors[index] -= ((self.learning_rate * np.sum(error_gradient, axis=1)) / self.num_train_data_points).reshape(self.bias_vectors[index].shape[0], 1)

            error_gradient = np.dot(self.weight_matrices[index].T, error_gradient) * NN.ReLU_derivative(weighted_sum_matrix)

        # print ("POST back prop weight matrices")
        # for i in self.weight_matrices:
        #     print (i, "\n")
        # print ("\n")


    def predict(self, should_print=True):

        # A = self.X_test

        forward_prop_prediction_inputs = [self.X_test] # training_set + activations
        
        # forward propagation process
        for index in range(self.num_hidden_layers):
            # Z = self.weight_matrices[index] @ A # could just do fowrard_prop_inputs[-1], but i'll keep track using "index" for now
            Z = self.weight_matrices[index] @ forward_prop_prediction_inputs[index]
            Z += self.bias_vectors[index] # Z = linear sum --> (W @ X) + b
            A = NN.ReLU(Z) # A = activation of Z
            forward_prop_prediction_inputs.append(A)
        # for the output layer
        Z = self.weight_matrices[-1] @ forward_prop_prediction_inputs[-1]
        Z += self.bias_vectors[-1]
        self.output_probability_matrix = NN.softmax(Z.T, axis=1)

        self.output_classes = np.argmax(self.output_probability_matrix, axis=1)

        self.correct_predictions = self.output_classes == self.Y_test
        self.accuracy = sum(self.correct_predictions) / self.num_test_data_points

        self.accuracies_over_all_epochs = np.append(self.accuracies_over_all_epochs, self.accuracy)

        if should_print:
            print (f"accuracy: {self.accuracy}")

        
        
        
        


    # Helper Functions
    
    def ReLU(Z):
        return np.maximum(Z, 0)

    def ReLU_derivative(Z):
         return (Z > 0).astype(float)

    def softmax(Z, axis):
        # Z = Z - np.max(matrix, axis=1, keepdims=True)
        # return np.exp(Z) / np.sum(np.exp(Z), axis=1, keepdims=True)
        max_val = np.max(Z, axis=axis, keepdims=True)
        exponents = np.exp(Z - max_val) # to keep stable and prevent overflow
        return exponents / np.sum(exponents, axis=axis, keepdims=True)
    
    def one_hot(Y):
        Y = Y.astype(int)
        # one_hot_Y = np.zeros(( Y.size, (np.max(Y) - np.min(Y) + 1) ))   # = array with dimensions: [ (n) , (# of classes, which is 10) ]
        one_hot_Y = np.zeros(( Y.size, len(np.unique(Y)) ))   # = array with dimensions: [ (n) , (# of classes, which is 10) ]
        one_hot_Y[np.arange(Y.size), Y - np.min(Y)] = 1
        return one_hot_Y
```
