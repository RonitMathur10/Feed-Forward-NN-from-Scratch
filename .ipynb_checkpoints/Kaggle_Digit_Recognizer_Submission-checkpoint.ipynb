# Imports

```python
%load_ext autoreload
%autoreload 2
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from datetime import datetime

from Feed_Forward_NN_Class import NN
```

# NN-Wrapper Class

```python
class model(NN):
    def __init__(self, X_train: np.ndarray, Y_train: np.ndarray, X_test: np.ndarray, Y_test: np.ndarray, hidden_layer_sizes: np.ndarray):
        super().__init__(X_train, Y_train, X_test, Y_test, hidden_layer_sizes)

    def save_model(self):
        return
```

# Digits Data

## Train Data

```python
digit_train_data = pd.read_csv("Data/digit-recognizer/train.csv")
print (digit_train_data.shape)
digit_train_data.head(5)
```

```python
digit_train, digit_test = digit_train_data.iloc[:, 1:].to_numpy(), digit_train_data.iloc[:, 0].to_numpy()
digit_X_train, digit_X_test, digit_Y_train, digit_Y_test = train_test_split(digit_train, digit_test, test_size=0.2, random_state=42, stratify=digit_test)
print (digit_X_train.shape, "\n", digit_Y_train.shape)
```

```python
# standardizing data
scaler = StandardScaler()
digit_X_train_scaled = scaler.fit_transform(digit_X_train)
# digit_X_test_scaled = scaler.fit_transform(digit_X_test)
digit_X_test_scaled = scaler.transform(digit_X_test)
```

```python
digit_X_train_scaled = digit_X_train_scaled.T # shape now = [classes x n]
digit_Y_train = digit_Y_train.T # shape now = [1 row x n cols]
digit_X_test_scaled = digit_X_test_scaled.T # shape now = [classes x n]
digit_Y_test = digit_Y_test.T # shape now = [1 row x n cols]
print (digit_X_train_scaled.shape, "\n", digit_Y_train.shape)
```

## Test Data

```python
digit_test_data = pd.read_csv("Data/digit-recognizer/test.csv")
print (digit_test_data.shape)
digit_test_data.head(5)
```

```python
digit_test_data = digit_test_data.to_numpy()
print (digit_test_data.shape, "\n")
digit_test_data_scaled = scaler.transform(digit_test_data)
digit_test_data_scaled = digit_test_data_scaled.T
print (np.mean(digit_test_data_scaled), digit_test_data_scaled.shape)
```

# Training Models

```python
# base_model = NN(X_train=digit_X_train_scaled, Y_train=digit_Y_train, X_test=digit_X_test_scaled, Y_test=digit_Y_test, hidden_layer_sizes=np.array([50, 25]))
base_model = model(X_train=digit_X_train_scaled, Y_train=digit_Y_train, X_test=digit_X_test_scaled, Y_test=digit_Y_test, hidden_layer_sizes=np.array([50, 25]))
```

```python
base_model.train(epochs=150, learning_rate=0.1)
```

```python
base_model.train(epochs=300, learning_rate=1)
```

# Prediction

```python
base_model_predictions = base_model.predict(X_data=digit_test_data_scaled)
base_model_predictions.shape
```

```python
# ---Kaggle Submission---

# flatten and ensure predictions are integers (0 instead of 0.0)
final_predictions = np.asarray(base_model_predictions).flatten().astype(int)

# create df and shift the index to start at 1 (kaggle required format)
df = pd.DataFrame({'Label': final_predictions})
df.index = df.index + 1

# save to a CSV
curr_time = datetime.now().strftime("%Y-%m-%d_%H-%M-%S")
df.to_csv(f"/Users/ronitmathur/Desktop/ML/Feed-Forward_NN/Kaggle/base_model_submission_{curr_time}.csv", index_label='ImageId')
```
